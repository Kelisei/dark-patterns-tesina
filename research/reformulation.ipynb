```python

from transformers import pipeline
import pandas as pd

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")


def reformulate_shaming(text):
    text_clean = text.strip()
    text_lower = text_clean.lower()
    
    if text_lower.startswith("no,") or text_lower.startswith("no gracias,"):
        return "no"
    elif text_lower.startswith("si,") or text_lower.startswith("sí,"):
        return "sí"
    
  
    try:
        result = classifier(text_clean)[0] 
        label = result['label']
        if label == "POSITIVE":
            return "sí"
        elif label == "NEGATIVE":
            return "no"
    except Exception as e:
        print("Error en clasificación:", e)
    
    
    return text_clean  


data = pd.read_csv("datasets/unified_dataset.csv")  
shaming = data[data['type'] == 'shaming']["content"]



for ex in shaming:
    reformulated = reformulate_shaming(ex)
    print(f"Original: {ex}  -> Reformulado: {reformulated}")
```
